# What is MCP?

**MCP (Model Context Protocol)** is a standardized protocol that allows an **AI application/LLM to connect with external tools, data, and services** in a consistent way.

For example, an LLM by itself cannot directly:

- Read your GitHub repositories
- Create GitHub issues
- Search your database
- Access files on your computer
- Query a company's internal API

MCP provides a **standard way for the LLM to interact with these external systems.**

---

# The Problem

Imagine we have:

```text
AI Application
      |
      |---- GitHub API
      |
      |---- Slack API
      |
      |---- Google Drive API
      |
      |---- Jira API
      |
      |---- Database API
```

This becomes complicated very quickly.

And there is another problem.

Suppose **10 different AI applications** want to integrate with GitHub.

Without MCP:

```text
AI App 1 ─────── GitHub API
AI App 2 ─────── GitHub API
AI App 3 ─────── GitHub API
AI App 4 ─────── GitHub API
...
AI App 10 ────── GitHub API
```

Every application has to build and maintain its own GitHub integration.

---

# MCP's Solution

MCP introduces a **standard interface** between AI applications and external services.

Instead of every AI application directly implementing the GitHub integration, we can have a:

**GitHub MCP Server**

```text
AI Application
      |
      | MCP
      ↓
GitHub MCP Server
      |
      | GitHub API
      ↓
    GitHub
```

The important point is:

> **MCP does not replace the GitHub API.**

The GitHub MCP server still uses GitHub's API internally.

MCP simply provides a **standardized interface for AI applications to interact with GitHub.**

---

# Example: Asking the AI to Find GitHub Issues

Suppose you ask:

> "Find all open issues in my repository."

The LLM itself doesn't know how to directly access GitHub.

The MCP client connects to a GitHub MCP server.

The GitHub MCP server might expose tools such as:

```text
get_repository
list_issues
create_issue
get_pull_request
create_pull_request
search_code
```

The AI can discover these tools through MCP.

For example:

```text
User
 ↓
"Find open issues in my repository"
 ↓
LLM
 ↓
MCP Client
 ↓
GitHub MCP Server
 ↓
GitHub API
 ↓
GitHub
```

The MCP server calls the appropriate GitHub API endpoint and returns the result.

The LLM then uses that result to answer the user.

---

# What MCP Actually Standardizes

The important thing is that MCP standardizes **how the AI communicates with the external tool**.

For example, the MCP server can tell the client:

```text
Tool: list_issues

Description:
List issues in a GitHub repository.

Arguments:
- owner
- repository
- state
```

The LLM can then decide:

> "I need to use the `list_issues` tool."

The MCP client sends the tool call to the MCP server using the MCP protocol.

The server handles the actual GitHub API interaction.

---

# Why This Is Powerful

Now imagine another AI application wants GitHub access.

It doesn't necessarily need to implement the entire GitHub integration itself.

It can connect to the existing GitHub MCP server:

```text
AI App A ──┐
AI App B ──┤
AI App C ──┼── MCP ──→ GitHub MCP Server ──→ GitHub
AI App D ──┤
AI App E ──┘
```

Similarly:

```text
                  ┌── GitHub MCP Server ──→ GitHub
                  |
AI Application ───┼── Slack MCP Server ───→ Slack
                  |
                  ├── Jira MCP Server ───→ Jira
                  |
                  └── Database MCP Server → Database
```

The AI application interacts with all of them through the **same MCP concept**, rather than having to build a completely different integration architecture for every service.

---

# In One Sentence

**MCP is a standardized protocol that allows AI applications to discover and use external tools and data sources through a common interface, while the MCP server handles the service-specific API communication.**

So the key distinction is:

## Without MCP

```text
AI Application → GitHub API
AI Application → Slack API
AI Application → Jira API
AI Application → Google API
```

The AI application must understand and implement each service's API.

## With MCP

```text
AI Application
      ↓
     MCP
      ↓
MCP Server
      ↓
Service's API
```

The **MCP server handles the service-specific details**, while the AI application gets a standardized way to interact with tools.

# MCP Transport Layer

A **transport protocol** in MCP defines **how the MCP Client and MCP Server communicate with each other**.

Think of MCP as the **language/protocol** they use, and the transport as the **communication channel** through which those messages travel.

For example, MCP may define a message such as:

```text
"Call the list_issues tool with repository = my-repo"
```

The transport determines **how this message gets from the Client to the Server and how the response comes back**.

```text
                    MCP
                     │
            Defines the messages
                     │
                     ▼
              Transport Layer
                     │
          ┌──────────┼──────────┐
          │          │          │
        STDIO       SSE    Streamable HTTP
          │          │          │
       Local       HTTP       HTTP
      processes   streaming   streaming
```

---

# 1. STDIO — Standard Input/Output

**STDIO** uses a program's standard input (`stdin`) and standard output (`stdout`) to communicate between the MCP Client and MCP Server.

It is primarily used when the MCP Server is running **locally on the same machine**.

## How it works

The MCP Client starts the MCP Server as a local process.

```text
┌──────────────────┐
│    MCP Client    │
│                  │
│    AI App        │
└────────┬─────────┘
         │
         │ stdin
         │ MCP request
         ▼
┌──────────────────┐
│    MCP Server    │
│  Local Process   │
└────────┬─────────┘
         │
         │ stdout
         │ MCP response
         ▼
      MCP Client
```

For example, the client might send:

```text
Client → Server:

Call list_issues(
    repository="my-repo"
)
```

The server processes it and writes the response to `stdout`:

```text
Server → Client:

[
    issue1,
    issue2,
    issue3
]
```

### Characteristics

* **Communication:** Bidirectional
* **Location:** Same machine
* **Complexity:** Very low
* **Network required:** No
* **Common use:** Local MCP servers

### Use cases

STDIO is ideal for:

* Local development and debugging
* CLI AI assistants
* Desktop AI applications
* Local filesystem MCP servers
* Local database MCP servers
* Development tools

For example:

```text
Claude Desktop
      │
      │ STDIO
      ▼
Filesystem MCP Server
      │
      ▼
Your computer's files
```

The important point is:

> **STDIO is essentially process-to-process communication on the same machine.**

---

# 2. SSE — Server-Sent Events

**SSE (Server-Sent Events)** is an HTTP-based mechanism where the **server can continuously push events to the client** over a persistent HTTP connection.

The important characteristic is:

> **Server → Client streaming**

```text
┌───────────────┐
│  MCP Client   │
└───────┬───────┘
        │
        │ HTTP request
        ▼
┌───────────────┐
│  MCP Server   │
└───────┬───────┘
        │
        │ SSE stream
        │
        ├──────────→ Event 1
        ├──────────→ Event 2
        ├──────────→ Event 3
        └──────────→ Event 4
```

The client establishes an HTTP connection and keeps it open.

The server can then continuously send events without the client having to repeatedly poll the server.

### Example

Imagine an MCP server performing a long-running operation:

```text
Client → Server:

Search arXiv for "transformers"
```

The server could send:

```text
Server → Client:

"Search started"

"Searching papers..."

"Found 10 papers"

"Search completed"
```

### Characteristics

* **Communication:** Primarily Server → Client streaming
* **Protocol:** HTTP
* **Connection:** Persistent
* **Complexity:** Moderate
* **Good for:** Streaming server events

### Use cases

SSE can be useful for:

* Streaming results
* Progress updates
* Status notifications
* Web-based applications

For example:

```text
MCP Client
    │
    │ HTTP
    ▼
MCP Server
    │
    ├── "Searching..."
    ├── "Found 5 results"
    ├── "Processing..."
    └── "Complete"
```

### Important MCP note

**SSE was used by earlier versions of MCP, but it is not the current recommended transport.**

Modern MCP uses **Streamable HTTP** for remote communication.

---

# 3. Streamable HTTP

**Streamable HTTP** is the modern HTTP transport for MCP and is designed for **remote MCP servers**.

It provides a standardized way for MCP messages to travel over HTTP while supporting streaming when necessary.

```text
┌──────────────────┐
│    MCP Client    │
│    AI App        │
└────────┬─────────┘
         │
         │ HTTP
         ▼
       Internet
         │
         ▼
┌──────────────────┐
│    MCP Server    │
│     Remote       │
└──────────────────┘
```

The important difference from the simplified SSE model is that **Streamable HTTP is not simply "WebSockets over HTTP."** It uses HTTP itself and can use streaming mechanisms such as SSE for server-to-client streaming when appropriate.

---

## How Streamable HTTP works

Suppose the user says:

> "Find all open issues in my GitHub repository."

The communication could look conceptually like:

```text
User
  │
  ▼
LLM
  │
  ▼
MCP Client
  │
  │ HTTP
  │
  │ "Call list_issues"
  ▼
MCP Server
  │
  │ GitHub API
  ▼
GitHub
  │
  ▼
MCP Server
  │
  │ HTTP response / streaming
  ▼
MCP Client
  │
  ▼
LLM
```

The MCP server handles the GitHub-specific API interaction.

---

# STDIO vs SSE vs Streamable HTTP

| Feature              | STDIO                                                   | Server-Sent Events (SSE)                                   | Streamable HTTP                                          |
| -------------------- | ------------------------------------------------------- | ---------------------------------------------------------- | -------------------------------------------------------- |
| **Primary Use**      | Local development, debugging, and local AI applications | Web-based server-to-client streaming; legacy MCP transport | Production, remote MCP servers, and scalable deployments |
| **Direction**        | Bidirectional                                           | Primarily Server → Client streaming                        | Bidirectional                                            |
| **Communication**    | Process `stdin` / `stdout` streams                      | HTTP persistent connection with SSE                        | HTTP-based communication with optional streaming         |
| **Streaming**        | Possible                                                | Yes — continuous Server → Client events                    | Yes — supports streaming when needed                     |
| **Deployment**       | Local machine                                           | Web / Network                                              | Web / Network                                            |
| **Network Required** | ❌ No                                                    | ✅ Yes                                                      | ✅ Yes                                                    |
| **Complexity**       | Lowest                                                  | Moderate                                                   | Moderate                                                 |
| **MCP Status**       | Current choice for local servers                        | **Legacy** transport                                       | **Recommended modern remote transport**                  |
| **Scalability**      | Limited to local process                                | Suitable for web streaming                                 | Designed for scalable remote deployments                 |

# How Transport Fits Into MCP

It is useful to visualize the entire architecture as layers:

```text
┌──────────────────────────────────────┐
│              LLM / AI                │
└──────────────────┬───────────────────┘
                   │
                   ▼
┌──────────────────────────────────────┐
│            MCP Client                │
└──────────────────┬───────────────────┘
                   │
                   │ MCP messages
                   ▼
┌──────────────────────────────────────┐
│          TRANSPORT LAYER             │
│                                      │
│     STDIO / SSE / Streamable HTTP    │
└──────────────────┬───────────────────┘
                   │
                   ▼
┌──────────────────────────────────────┐
│            MCP Server                │
└──────────────────┬───────────────────┘
                   │
                   ▼
          External Service
        GitHub / Database /
        Files / APIs / etc.
```

The key idea is:

> **The transport does not define what a tool does. It only defines how MCP messages travel between the Client and Server.**

For example, the MCP server might expose:

```text
list_issues()
create_issue()
search_code()
```

Those tools don't fundamentally change depending on the transport.

With STDIO:

```text
MCP Client
    │
    │ stdin/stdout
    ▼
MCP Server
```

With Streamable HTTP:

```text
MCP Client
    │
    │ HTTP
    ▼
MCP Server
```

The **MCP protocol remains the same; the communication mechanism changes.**

---

# Why Do We Need Different Transports?

Because MCP can be used in very different environments.

### Local AI application

```text
AI App
  │
  │ STDIO
  ▼
Local MCP Server
```

No network infrastructure is necessary.

### Remote AI application

```text
AI App
  │
  │ Streamable HTTP
  ▼
Remote MCP Server
```

The server can be hosted on a remote machine/cloud.

This makes it possible for multiple clients to access a remotely hosted MCP server.

---

# The Big Picture

Think about it like this:

```text
MCP
│
├── Defines WHAT is communicated
│
│   ├── Tools
│   ├── Resources
│   ├── Prompts
│   └── JSON-RPC messages
│
└── Transport defines HOW it is communicated
    │
    ├── STDIO
    │      └── Local process communication
    │
    ├── SSE
    │      └── HTTP server → client streaming
    │      └── Legacy MCP transport
    │
    └── Streamable HTTP
           └── Modern remote MCP transport
           └── HTTP-based communication
           └── Supports streaming
```

### In one sentence:

> **MCP defines the standardized communication protocol between an MCP Client and Server, while the transport layer determines how those MCP messages are physically exchanged—locally through STDIO or remotely through HTTP-based transports such as Streamable HTTP.**


# MCP's 3 Core Data Primitives

The **three core primitives** of MCP are:

> **Tools → Actions**
> **Resources → Data**
> **Prompts → Instructions**

They describe the three fundamental things an **MCP Server can expose to an MCP Client/AI application**.

```text
                         MCP SERVER
                              │
                 ┌────────────┼────────────┐
                 │            │            │
                 ▼            ▼            ▼
              TOOLS       RESOURCES     PROMPTS
                 │            │            │
                 ▼            ▼            ▼
              ACTIONS        DATA       INSTRUCTIONS
```

Let's understand each one properly.

---

# 1. Tools — "What can the AI DO?"

A **Tool** is an operation/function that the AI can invoke to **perform an action**.

Think of a tool as a normal programming function:

```python
def create_issue(repository, title, description):
    ...
```

The MCP server exposes this function as a **tool**.

For example, a GitHub MCP server could expose:

```text
Tools
│
├── list_issues()
├── create_issue()
├── search_code()
├── create_pull_request()
└── merge_pull_request()
```

The LLM can decide:

> "I need to create a GitHub issue, so I should call `create_issue`."

The flow becomes:

```text
User
 │
 │ "Create an issue saying the login page is broken."
 ▼
LLM
 │
 │ decides to call create_issue
 ▼
MCP Client
 │
 │ tool call
 ▼
MCP Server
 │
 │ executes create_issue()
 ▼
GitHub API
 │
 ▼
GitHub Issue Created
```

### What are Tools used for?

Tools are used whenever the AI needs to **do something**.

Examples:

```text
GitHub:
    create_issue()
    create_pull_request()

Database:
    run_query()

Email:
    send_email()

Filesystem:
    create_file()

Calculator:
    calculate()
```

### Important point

Tools can have **side effects**.

For example:

```text
create_issue()
delete_file()
send_email()
merge_pull_request()
```

These actually change something in the external system.

So remember:

> **Tool = an action/function that the AI can invoke.**

---

# 2. Resources — "What information can the AI READ?"

A **Resource** represents **data or information** that an MCP server makes available to the AI application.

Think of a resource like a **file, document, database record, API response, or other piece of contextual information**.

For example, a GitHub MCP server could expose resources such as:

```text
Resources
│
├── Repository README
├── package.json
├── Source code
├── Issue information
└── Documentation
```

The important distinction is:

> **Tools perform actions; Resources provide information.**

For example:

```text
Tool:
    create_issue()
        ↓
    DOES something

Resource:
    README.md
        ↓
    PROVIDES information
```

---

## Example

Suppose you ask:

> "What dependencies does this GitHub project use?"

The AI needs information from `package.json`.

The MCP server could expose:

```text
github://repo/my-project/package.json
```

The AI application can retrieve that resource and provide it as context to the LLM.

```text
             MCP Server
                  │
                  │ Resource
                  ▼
          package.json
                  │
                  ▼
             MCP Client
                  │
                  ▼
                 LLM
                  │
                  ▼
       "This project uses React,
        Next.js and TypeScript."
```

### What are Resources used for?

Resources are useful for providing:

* Files
* Documents
* Database records
* Repository information
* Configuration
* Documentation
* API data
* Logs
* Other contextual information

So:

> **Resource = data/context that the AI application can read and use.**

---

# 3. Prompts — "How should the AI perform a task?"

A **Prompt** is a predefined, reusable **instruction/template** exposed by the MCP server.

This is slightly different from a normal user message.

For example, suppose your GitHub MCP server provides:

```text
review_pull_request
```

The prompt might be designed like:

```text
Review the following pull request.

Analyze:
1. Bugs
2. Security vulnerabilities
3. Performance problems
4. Code quality
5. Potential improvements
```

The application can use this predefined prompt rather than constructing the instructions itself every time.

---

## Example

Imagine a GitHub MCP server exposes:

```text
Prompts
│
├── review_pull_request
├── summarize_repository
└── explain_issue
```

The user could select:

```text
"Review this pull request."
```

The application retrieves the predefined prompt:

```text
review_pull_request(PR #123)
```

which gives the LLM structured instructions for performing the review.

```text
User
 │
 │ "Review PR #123"
 ▼
MCP Client
 │
 │ gets prompt
 ▼
MCP Server
 │
 │ review_pull_request
 ▼
Structured instructions
 │
 ▼
LLM
 │
 ▼
Code Review
```

So:

> **Prompt = reusable instructions for guiding the LLM.**

---

# The Most Important Difference

This is where people commonly get confused.

Imagine a GitHub MCP server:

```text
                    GitHub MCP Server
                           │
              ┌────────────┼────────────┐
              │            │            │
              ▼            ▼            ▼
            TOOLS       RESOURCES     PROMPTS
              │            │            │
              ▼            ▼            ▼
           "DO"          "READ"       "GUIDE"
```

### Tools

**Do something**

```text
create_issue()
merge_pr()
delete_branch()
```

### Resources

**Give information**

```text
README.md
package.json
issue #123
source_code
```

### Prompts

**Give instructions**

```text
review_pull_request
summarize_repository
explain_issue
```

# A Complete GitHub Example

Suppose you tell your AI:

> **"Look at PR #42, review the code, and if you find a serious bug, create an issue."**

Now all three primitives can potentially be involved.

### Step 1 — Resource

The AI needs to **read the PR/code**.

```text
Resource
    ↓
PR #42
    ↓
Source code / changes
    ↓
LLM
```

### Step 2 — Prompt

The AI can use a predefined review prompt:

```text
Prompt
    ↓
review_pull_request
    ↓
"Analyze bugs, security, performance..."
    ↓
LLM
```

### Step 3 — Tool

The AI finds a serious bug and decides to create an issue:

```text
Tool
    ↓
create_issue()
    ↓
GitHub API
    ↓
Issue created
```

So the complete flow could look like:

```text
                         GitHub MCP Server
                                │
              ┌─────────────────┼─────────────────┐
              │                 │                 │
              ▼                 ▼                 ▼
          RESOURCES           PROMPTS           TOOLS
              │                 │                 │
              ▼                 ▼                 ▼
            PR #42       review_pull_request   create_issue()
              │                 │                 │
              └────────────┬────┘                 │
                           ▼                      │
                          LLM                     │
                           │                      │
                           │ finds serious bug    │
                           └──────────────────────┘
                                      │
                                      ▼
                                GitHub Issue
```

# Who Controls Each Primitive?

This distinction is also important in MCP.

Generally:

| Primitive     | Main purpose                  | Control                    |
| ------------- | ----------------------------- | -------------------------- |
| **Tools**     | Perform actions               | **Model-controlled**       |
| **Resources** | Provide data/context          | **Application-controlled** |
| **Prompts**   | Provide reusable instructions | **User-controlled**        |

### Model-controlled → Tools

The model can decide:

> "I need to call `search_code`."

### Application-controlled → Resources

The host application decides what resource/context should be made available to the model.

### User-controlled → Prompts

The user typically chooses which predefined prompt/template they want to use.

```text
Documents
Customer records
Reports
Logs
```

```text
"Review this document"
"Summarize this report"
"Analyze this issue"
```

Therefore:

```text
             MCP SERVER
                 │
       ┌─────────┼─────────┐
       │         │         │
       ▼         ▼         ▼
     TOOLS   RESOURCES   PROMPTS
       │         │         │
      DO        READ      GUIDE
       │         │         │
       ▼         ▼         ▼
    Actions     Data    Instructions
```

### The one-line definition to remember

> **MCP Tools allow the model to perform actions, Resources provide data/context to the AI application, and Prompts provide reusable instructions/templates for guiding the model.**
