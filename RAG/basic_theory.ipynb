**Retrieval-Augmented Generation (RAG)** is the process of optimizing the **output of a large language model**, so it references an **authoritative knowledge base** outside of its **training data sources** before generating a response. **Large Language Models (LLMs)** are trained on vast volumes of data and use billions of parameters to generate original output for tasks like answering questions, translating languages, and completing sentences. 
One Problem it faces is that it doesnt have any information outside of its training data whihc include company specific private data, or all the data uploaded on the internet uptill the LLM's training date. To give the LLM knowledge about the data it doesnt know there are 2 methods-
1. **Fine-Tune** the LLM on the data you want it to have, but its downside is that fine-tuning a model with billions of parameters (even after freezing the ending layers) is very expensive i.e. needs lots o computational power, therefore we usually use the second method given below
2. **RAG** extends the already powerful capabilities of **LLMs** to specific domains or an **organization's internal knowledge base**, all without the need to retrain the model. It is a cost-effective approach to improving **LLM** output so it remains relevant, accurate, and useful in various contexts.

# Retrieval-Augmented Generation (RAG) Architecture

---

## 1. Pipelines Overview

### **Data Ingestion Pipeline** *(Preparation Phase)*

* **Data Collection & Parsing:** External data sources (PDFs, HTML, Excel, SQL) are ingested. Parsing converts these raw, unstructured file formats into clean plain text and breaks them into smaller, manageable segments (*chunks*).
* **Embedding Creation:** A specialized embedding model processes these text chunks and transforms them into numerical mathematical arrays called *vectors*.
* **Storage:** These vector representations are saved into a **Vector Database** (Vector Store) to build a searchable knowledge base.

### **Inference Pipeline** *(Runtime / Query Phase)*

* **User Query:** A user asks a question (e.g., *"What is the leave policy?"*).
* **Query Embedding:** The user's text question is converted into a vector using the same embedding model used during ingestion.
* **Retrieval:** The **Retriever** performs a similarity search in the Vector Database to find text chunks whose vector embeddings are closest in meaning to the query vector.
* **Augmentation & Generation:** The relevant retrieved text chunks (*Context*) and the original user question (*Prompt*) are sent together to the **LLM**, which generates an accurate, context-aware answer.

## 2. Core Concepts

### **What is Parsing?**

> **Parsing** is the technical step of extracting raw text, structured tables, or metadata out of non-plain-text document formats (such as complex PDFs, HTML layouts, slide decks, or spreadsheets) and organizing it so it can be cleanly chunked and embedded without garbage code or broken formatting.
