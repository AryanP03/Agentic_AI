STREAMING
Most models can stream their output content while it is being generated. By displaying output progressively,
 streaming significantly improves user experience, particularly for longer responses. Calling stream() returns an iterator that 
 yields output chunks as they are produced. You can use a loop to process each chunk in real-time:




```python
response =model.invoke("Write a 200 word para on AI")
response.content
# w/o streaming we need to wait for the entire message, as it comes at once after waiting
```

```python
for chunk in model.stream("Write a 200 word essay on AI"):
    print(chunk.text)
```
