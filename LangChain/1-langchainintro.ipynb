```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.agents import create_agent

def get_weather(city: str) -> str: # this is just one tool, we can have multiple tools here
    """Get the weather for a city"""
    return f"The weather in {city} is sunny"

llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash"
)

agent = create_agent(
    model=llm,
    tools=[get_weather],
    system_prompt="You are a helpful assistant"
)
```

#### this is an autonomous agent, since LLM calls the tool not us, decision is made by LLM(gemini here) 
The LLM now reasons: "I don't know the live weather, but I have a tool called get_weather whose description matches this request."

```python
response = agent.invoke(
    {
        "messages": [
            {
                "role": "user",
                "content": "What is the weather in Pune?"
            }
        ]
    }
)
#Since we are using the inbuilt function of create_agent above we need to give the input in the form of a dictionary where dictionary key will be-"messages"

print(response)
# tool_call_id='723b390b-abe0-4792-89be-8a9a264e3f0a') tool call made

response["messages"][-1].content
#this is for printing the last output i.e. the final output sent by the llm

```
